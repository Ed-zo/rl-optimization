```python
import numpy as np

# Load the data from the file
data = np.loadtxt('dist.txt')

#conver to int
data = data.astype(int)
```

```python
tij = np.zeros((115, 115))

for n in range(len(data)):
    i = data[n][0] - 1
    j = data[n][1] - 1
    tij[i][j] = data[n][2]
    tij[j][i] = data[n][2]

print(tij)
```

```python
import pandas as pd

df = pd.read_csv('tasks.csv', header=0, sep=';')

t = np.zeros((725, 725))

for i in range(len(df)):
    finish = df['ZastavkaFinish'][i]
    for j in range(len(df)):
        start = df['ZastavkaStart'][j]
        t[i][j] = tij[finish][start]

print(t)
```
